## Exercise 1
### Which linear regression training algorithm can you use if you have a training set with millions of features?
we can use Stochastic Gradient Descent (SGD) or Mini-batch Gradient Descent. Batch Gradient Descent also works, but algorithms that use the Normal Equation or SVD are far too slow because their computational complexity grows exponentially as the number of features increases.

## Exercise 2
### Suppose the features in your training set have very different scales. Which algo‐ rithms might suffer from this, and how? What can you do about it?
Gradient Descent algorithms suffer heavily from unscaled features. The cost function becomes a highly elongated bowl, meaning the algorithm will take a very long time to converge as it zig zags down the valley. You can fix this by scaling the data before training (for example, using Scikit-Learn’s StandardScaler).

## Exercise 3
### Can gradient descent get stuck in a local minimum when training a logistic regression model?
No, Gradient Descent cannot get stuck in a local minimum when training a logistic regression model. This is because its cost function is convex (shaped like a bowl), meaning it only has one single global minimum.


## Exercise 4
### Do all gradient descent algorithms lead to the same model, provided you let them run long enough?
Not exactly. If the problem is convex and the learning rate isn't too high, they will all approach the global minimum. However, unless you gradually reduce the learning rate over time, Stochastic and Mini-batch Gradient Descent will never actually settle down; they will keep bouncing around the very bottom. So, Batch GD will find the exact minimum, while the others will end up slightly different.

## Exercise 5
### Suppose you use batch gradient descent and you plot the validation error at every epoch. If you notice that the validation error consistently goes up, what is likely going on? How can you fix this?
If both the validation error and the training error are going up, your learning rate is too high, and the algorithm is diverging. You fix this by lowering the learning rate. However, if the training error is going down while the validation error goes up, your model is overfitting the training data, and you should stop training (Early Stopping).

## Exercise 6
### Is it a good idea to stop mini-batch gradient descent immediately when the validation error goes up?
No, this is a bad idea. Because Mini-batch and Stochastic GD use random subsets of data, their path down the gradient is erratic. The validation error will bounce up and down naturally. If you stop immediately at the first bounce upward, you will likely stop way too early. Instead, save the model at regular intervals and stop only when the error has been trending upward for a while (then revert to the best saved model).

## Exercise 7
### Which gradient descent algorithm (among those we discussed) will reach the vicinity of the optimal solution the fastest? Which will actually converge? How can you make the others converge as well?
Fastest to the vicinity: Stochastic Gradient Descent (SGD) is the fastest to get close to the optimal solution because it updates the model parameters after looking at just one single training instance.

Actually converges: Batch Gradient Descent is the only one that will stop exactly at the true minimum, given enough time.

How to make others converge: You must use a learning schedule to gradually reduce the learning rate as the algorithm gets closer to the minimum, which forces SGD and Mini-batch GD to settle down instead of bouncing around.
